```python
from sklearn.preprocessing import StandardScaler

df = pd.DataFrame(df)

numeric_cols = df.select_dtypes(include=[np.number]).columns
X = df[numeric_cols]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

df_scaled = pd.DataFrame(X_scaled, columns=numeric_cols)
print(df_scaled.describe())
```

```python
df_scaled.describe()
```

```python
df_scaled
```

```python
df_scaled = df_scaled.drop(['popularity_songs' , 'explicit','key','mode','time_signature','followers', 'popularity_artists'],axis=1 )
```

```python
df_scaled
```

```python
#finding k by elbow method

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score,davies_bouldin_score
from sklearn.decomposition import PCA

inertias = []
k_range = range(2,9)

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10, max_iter=300 )
    labels = kmeans.fit_predict(df_scaled)
    inertias.append(kmeans.inertia_)
    
```

```python
inertias
```

```python
plt.figure(figsize= (12,5))
plt.plot(k_range,inertias)
```

```python
#davies_bouldin_score

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score,davies_bouldin_score
from sklearn.decomposition import PCA


db_scores = []
k_range = range(2,9)

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10, max_iter=300 )
    labels = kmeans.fit_predict(df_scaled)
    db_scores.append(davies_bouldin_score(df_scaled, labels))
```

```python
db_scores
```

```python
plt.figure(figsize= (12,5))
plt.plot(k_range,db_scores)
```

```python
#calinski_harabasz_scores

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score,davies_bouldin_score,calinski_harabasz_score
from sklearn.decomposition import PCA

calinski_harabasz_scores = []
k_range = range(2,20)

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10, max_iter=300 )
    labels = kmeans.fit_predict(df_scaled)
    calinski_harabasz_scores.append(calinski_harabasz_score(df_scaled, labels))
    
```

```python
calinski_harabasz_scores
```

```python
plt.figure(figsize= (12,5))
plt.plot(k_range,calinski_harabasz_scores)
```

```python
#silhouette_scores

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score,davies_bouldin_score
from sklearn.decomposition import PCA


silhouette_scores = []

k_range = range(2,7)

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10, max_iter=300 )
    labels = kmeans.fit_predict(df_scaled)
    silhouette_scores.append(silhouette_score(df_scaled, labels, sample_size=50000))
```

```python
silhouette_scores
```

```python
kmeans_final = KMeans(n_clusters=3, random_state=42)
y_kmeans = kmeans_final.fit_predict(df_scaled)

a = silhouette_score(df_scaled, y_kmeans)
b = calinski_harabasz_score(df_scaled, y_kmeans)
c = davies_bouldin_score(df_scaled, y_kmeans)
```

```python
a,b,c
```

```python
# Calculate mean of each feature per cluster

cluster_profiles = df.groupby('cluster')[numeric_features.columns].mean()

cluster_profiles
```

cluster 0 --> “Chill acoustic / instrumental tracks”
cluster 1 --> “Party / Mainstream Hit Tracks”
cluster 2 --> “Speech-Heavy / Short-Form Experimental Tracks”

```python
cluster_labels = {
    0: "Chill acoustic / instrumental tracks" ,
    1: "Party / Mainstream Hit Tracks",
    2: "Speech-Heavy / Short-Form Experimental Tracks"
}

df['cluster_label'] = df['cluster'].map(cluster_labels)
```
